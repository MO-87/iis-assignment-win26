### ANSWERS
------------------------------------------
#### E1:

The robot moved smoothly and got the mug onto the table.. so it reached the visible end state. To solve the Body Motion Problem.. it also had to serve the drink without unwanted side effects throughout the task.

Two constraints were violated:

1. **Keep the milk contained..** milk spilled across the counter during the task
2. **Close the fridge after using it..** the robot left it open when it finished which shouldve been closed


#### E2:

- **Working interpretation** :: `(:type serving) (:recipient Eva) (:substance coffee-with-milk)` .. her usual mug already has black coffee in it.. so coffee with milk is a plausible interpretation.. the robot should check with Eva before adding anything
- **Inferred means** :: `(:vessel-role contain-and-support)` .. the drink needs a vessel to hold it while it’s carried and served.. Eva’s mug fills that role
- **Grounded physical object** :: `(:object carton-A)` .. once whole milk is chosen.. the robot uses the label to identify the opened whole-milk carton in the fridge door


#### E3:

- **5 mm test lift..** the robot checks whether its grip can hold Carton A while it’s still close to the shelf.. if the carton slips, the robot can set it back down and adjust its grip
- **High carry..** a stable grip gets the carton to the counter in one move.. if it slips at that height.. the carton can fall and spill milk

The evidence for choosing the low test lift is "the wet surface on Carton A".. it makes the side pinch more likely to slip.. the low lift checks the grip while a slip is still easy to recover from


#### E4:

- **Milk into coffee..** `(:stop-when (>= poured-volume 26.5 mL))` :: The target is 30 mL.. the robot commands the wrist to tilt back at 26.5 mL because the sensor reading is slightly delayed and some milk still drains from the spout
- **Pancake batter into a pan..** `(:stop-when (>= disc-diameter 9 cm))` :: The target is a 12 cm pancake.. the robot tilts the wrist back when the disc is about 9 cm across because the batter already in the pan keeps spreading after the flow stops


#### E5:

1. **Infer..** the robot works out that it needs whole milk to prepare Eva’s coffee with milk
2. **Ground..** it reads the labels and identifies Carton A as the whole milk
3. **Predict..** the side pinch might slip on the wet carton.. then a 5 mm lift keeps it close enough to the shelf to set it down if it slips
4. **Decide..** it chooses the low test lift before committing to the carry
5. **Regulate..** it lifts slowly and monitors the carton’s position relative to the gripper
6. **Detect & Revise..** downward motion "relative to the gripper" triggers this step. The robot detects that the side pinch is slipping and sets the carton down.. it revises the designator from `(:grasp side-pinch)` to `(:grasp (:type bottom-support) (:hand palm-under))`.
7. **Verify..** it tests the revised grasp and checks that the carton stays stable without slipping or denting before carrying it

#### E6:

A specific variant would be a nearly full Carton A with condensation on its sides and the cap off.. with the fixed high side pinch lift.. the wet surface makes the grip more likely to slip and the heavier carton is harder to hold.. if it slips during the carry.. it can fall and spill milk

The same fixed grip could work once on a dry carton.. that tells us what happened in that run. I think evidence of competence would come from testing dry and wet cartons at different fill levels and checking whether the robot detects when its grip is unreliable.. adapts it and completes the task without spills
